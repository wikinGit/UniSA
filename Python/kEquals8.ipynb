```python
# Simple voting app using a dictionary to store votes
votes = {}

def vote(candidate):
    if candidate in votes:
        votes[candidate] += 1
    else:
        votes[candidate] = 1

def show_results():
    for candidate, count in votes.items():
        print(f"{candidate}: {count} votes")

# Example usage:
vote("Alice")
vote("Bob")
vote("Alice")
show_results()
```
